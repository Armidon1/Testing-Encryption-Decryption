# Encryption Performance Analysis

This notebook provides an exploratory data analysis of the performance differences between various encryption algorithms. It visualizes the encryption and decryption times for different file sizes.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style of the plots
sns.set_style('whitegrid')

# Load the benchmark data
benchmark_data = pd.read_csv('../data/processed/benchmark_results.csv')

# Display the first few rows of the dataset
benchmark_data.head()
```

```python
# Plotting encryption times
plt.figure(figsize=(12, 6))
sns.lineplot(data=benchmark_data, x='file_size', y='encryption_time', hue='algorithm', marker='o')
plt.title('Encryption Time by File Size and Algorithm')
plt.xlabel('File Size (bytes)')
plt.ylabel('Encryption Time (seconds)')
plt.legend(title='Algorithm')
plt.grid(True)
plt.savefig('../data/processed/encryption_time_plot.png')
plt.show()
```

```python
# Plotting decryption times
plt.figure(figsize=(12, 6))
sns.lineplot(data=benchmark_data, x='file_size', y='decryption_time', hue='algorithm', marker='o')
plt.title('Decryption Time by File Size and Algorithm')
plt.xlabel('File Size (bytes)')
plt.ylabel('Decryption Time (seconds)')
plt.legend(title='Algorithm')
plt.grid(True)
plt.savefig('../data/processed/decryption_time_plot.png')
plt.show()
```
